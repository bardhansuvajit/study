```python
## Code to show all prime numbers from 1 to 100
numbers="Prime numbers: "
for i in range(2, 101):
    for j in range(2, i):
        if (i % j == 0):
            break
    else:
        numbers += str(i) + ", "
        # print(f"{i}")
print(numbers)
    
```

```python
squares = []
for i in range(1, 10):
    squares.append(i ** 2)
print(squares)
```

```python
squares = [i ** 2 for i in range(1, 10)]
print(squares)
```

```python
evens = [i for i in range(1, 10) if i % 2 == 0]
print(evens)
```
